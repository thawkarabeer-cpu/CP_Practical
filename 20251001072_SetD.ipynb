```python
# Part A: Batch reactor simulation and batch tracking
import math

# 1) Euler method simulation for 12 min
C0 = 2.5
k = 0.04
dt = 1.0
max_time = 12

print("Part A: Batch reactor simulation using Euler method")
print("Time(min)\tConcentration (mol/L)\tConversion (%)")

C = C0
for t in range(0, max_time + 1):
    conversion = ((C0 - C) / C0) * 100
    print(f"{t}\t\t{C:.4f}\t\t{conversion:.2f}")
    if t < max_time:
        C = C - k * C * dt
        if C < 0:
            C = 0.0

# 2) Exact solution using math.exp()
exact_time = 12
C_exact = C0 * math.exp(-k * exact_time)
C_euler = 2.5
for t in range(0, exact_time):
    C_euler = C_euler - k * C_euler * 1.0
    if C_euler < 0:
        C_euler = 0.0

absolute_error = abs(C_euler - C_exact)
print("\nExact concentration at 12 min:", round(C_exact, 6))
print("Euler concentration at 12 min:", round(C_euler, 6))
print("Absolute error:", round(absolute_error, 6))

# 3) While loop until conversion reaches 40% or time reaches 100 min
C2 = C0
t = 0.0
print("\nPart A3: Continue until conversion reaches 40% or time reaches 100 min")
while t <= 100 and ((C0 - C2) / C0) * 100 < 40:
    C2 = C2 - k * C2 * dt
    if C2 < 0:
        C2 = 0.0
    t += dt

conversion_final = ((C0 - C2) / C0) * 100
print(f"Time needed: {t:.1f} min")
print(f"Final concentration: {C2:.4f} mol/L")
print(f"Final conversion: {conversion_final:.2f}%")

# Part B: Distillation column sample codes
print("\nPart B: Distillation code digit analysis")

def digit_analysis(code):
    if code == "":
        print("Message: Empty code entered.")
        return

    digits = []
    total = 0
    for ch in code:
        if ch.isdigit():
            digits.append(ch)
            total += int(ch)

    if len(digits) == 0:
        print("Message: No digits found in the code.")
        return

    print("Digit count:", len(digits))
    print("Sum of digits:", total)
    print("Digits found:", ''.join(digits))

# Test cases
for sample in ["DIST2026X9Q3", "", "ABCDEF"]:
    print("\nCode tested:", sample)
    digit_analysis(sample)
```

```python
# This cell is left intentionally blank for extra work if needed.
import math

print("Equilibrium Conversion Checker")
print("Reaction: A + B ⇌ C")
print("Initial concentrations: [A]0 = 1.5 mol/L, [B]0 = 1.0 mol/L, [C]0 = 0")

raw_input = input("Enter Kc: ")
print("Raw input type:", type(raw_input).__name__)

try:
    Kc = float(raw_input)
    print("Converted value type:", type(Kc).__name__)

    if Kc <= 0:
        print("Invalid Kc: Kc must be greater than 0.")
    else:
        A0 = 1.5
        B0 = 1.0
        C0 = 0.0

        # Kc = x / ((1.5 - x)(1 - x))
        # Rearranged equation:
        # Kc*x^2 - (2.5*Kc + 1)*x + 1.5*Kc = 0
        a = Kc
        b = -(2.5 * Kc + 1)
        c = 1.5 * Kc

        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")

        discriminant = b**2 - 4*a*c
        print("Discriminant =", discriminant)

        if discriminant < 0:
            print("No real roots: discriminant is negative.")
        else:
            root1 = (-b + math.sqrt(discriminant)) / (2 * a)
            root2 = (-b - math.sqrt(discriminant)) / (2 * a)
            print("Root 1 =", root1)
            print("Root 2 =", root2)

            valid_roots = []
            if 0 <= root1 <= 1:
                valid_roots.append(root1)
            if 0 <= root2 <= 1:
                valid_roots.append(root2)

            if len(valid_roots) == 0:
                print("No physically meaningful root. Both roots are outside 0 <= x <= 1.")
            else:
                x = valid_roots[0]
                print("Physically meaningful root x =", x)

                A_eq = A0 - x
                B_eq = B0 - x
                C_eq = x

                conversion_percent = (x / B0) * 100

                print(f"Equilibrium [A] = {A_eq:.3f} mol/L")
                print(f"Equilibrium [B] = {B_eq:.3f} mol/L")
                print(f"Equilibrium [C] = {C_eq:.3f} mol/L")
                print(f"Conversion of limiting reactant B = {conversion_percent:.2f}%")

                if conversion_percent >= 70:
                    level = "High"
                elif conversion_percent >= 40:
                    level = "Moderate"
                else:
                    level = "Low"

                print(f"Classification: {level}")

except ValueError:
    print("Invalid input: please enter a numeric value for Kc.")
    print("No crash: the program handled the non-numeric input safely.")
```
