```python
# Part A: Solvent screening and sampling schedule

def factorial(n):
    if n == 0:
        return 1
    if n < 0:
        raise ValueError("n must be non-negative")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


def combination(n, r):
    if n < 0 or r < 0:
        print("Invalid input: n and r must be non-negative.")
        return None
    if r > n:
        print("Invalid input: r cannot be greater than n.")
        return None
    return factorial(n) // (factorial(r) * factorial(n - r))


print("Part A: Solvent screening")
print("Number of possible mixtures for n=9, r=3:", combination(9, 3))
print("Testing n=6, r=8:")
print(combination(6, 8))

# Part B: Fibonacci sampling schedule

def fib_iter(n):
    if n <= 0:
        return []
    fib = [0, 1]
    while len(fib) < n:
        fib.append(fib[-1] + fib[-2])
    return fib[:n]


def fib_rec(k):
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)

print("\nPart B: Fibonacci sampling schedule")
first_10 = fib_iter(10)
print("First 10 Fibonacci numbers:", first_10)

# Check agreement with recursive values
for i in range(10):
    if fib_rec(i) != first_10[i]:
        print("Mismatch at index", i)
        break
else:
    print("Recursive and iterative Fibonacci values agree for n=10.")

print("\nSample No. | Fibonacci tag | Sample time (min)")
for i, value in enumerate(first_10):
    print(f"{i+1:>9} | {value:>12} | {value * 4:>15}")
```

```python
import math

print("Equilibrium Conversion Checker")
print("Reaction: A + B ⇌ C")
print("Initial concentrations: [A]0 = 1.5 mol/L, [B]0 = 1.0 mol/L, [C]0 = 0")

raw_input = input("Enter Kc: ")
print("Raw input type:", type(raw_input).__name__)

try:
    Kc = float(raw_input)
    print("Converted value type:", type(Kc).__name__)

    if Kc <= 0:
        print("Invalid Kc: Kc must be greater than 0.")
    else:
        A0 = 1.5
        B0 = 1.0
        C0 = 0.0

        # Kc = x / ((1.5 - x)(1 - x))
        # Rearranged equation:
        # Kc*x^2 - (2.5*Kc + 1)*x + 1.5*Kc = 0
        a = Kc
        b = -(2.5 * Kc + 1)
        c = 1.5 * Kc

        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")

        discriminant = b**2 - 4*a*c
        print("Discriminant =", discriminant)

        if discriminant < 0:
            print("No real roots: discriminant is negative.")
        else:
            root1 = (-b + math.sqrt(discriminant)) / (2 * a)
            root2 = (-b - math.sqrt(discriminant)) / (2 * a)
            print("Root 1 =", root1)
            print("Root 2 =", root2)

            valid_roots = []
            if 0 <= root1 <= 1:
                valid_roots.append(root1)
            if 0 <= root2 <= 1:
                valid_roots.append(root2)

            if len(valid_roots) == 0:
                print("No physically meaningful root. Both roots are outside 0 <= x <= 1.")
            else:
                x = valid_roots[0]
                print("Physically meaningful root x =", x)

                A_eq = A0 - x
                B_eq = B0 - x
                C_eq = x

                conversion_percent = (x / B0) * 100

                print(f"Equilibrium [A] = {A_eq:.3f} mol/L")
                print(f"Equilibrium [B] = {B_eq:.3f} mol/L")
                print(f"Equilibrium [C] = {C_eq:.3f} mol/L")
                print(f"Conversion of limiting reactant B = {conversion_percent:.2f}%")

                if conversion_percent >= 70:
                    level = "High"
                elif conversion_percent >= 40:
                    level = "Moderate"
                else:
                    level = "Low"

                print(f"Classification: {level}")

except ValueError:
    print("Invalid input: please enter a numeric value for Kc.")
    print("No crash: the program handled the non-numeric input safely.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# a) Batch reactor data
# -----------------------------
data = {
    "Time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_l": [1.50, 1.17, 0.91, 0.71, 0.55, 0.43],
    "Run2_mol_l": [1.50, 0.96, 0.61, 0.39, 0.25, 0.16]
}

df = pd.DataFrame(data)

# Save to CSV and JSON without index column
df.to_csv("kinetics.csv", index=False)
df.to_json("kinetics.json", orient="records", indent=2)

# Read CSV back
df_read = pd.read_csv("kinetics.csv")

print("CSV data shape:", df_read.shape)
print(df_read.dtypes)

# -----------------------------
# b) Mean and standard deviation
# -----------------------------
run1_mean = df["Run1_mol_l"].mean()
run1_std = df["Run1_mol_l"].std()

run2_mean = df["Run2_mol_l"].mean()
run2_std = df["Run2_mol_l"].std()

print("\nRun1 mean:", run1_mean)
print("Run1 std:", run1_std)
print("Run2 mean:", run2_mean)
print("Run2 std:", run2_std)

if run2_std > run1_std:
    variation = "Run2_mol_l shows more variation."
else:
    variation = "Run1_mol_l shows more variation."

print("\n", variation)

# -----------------------------
# c) Plot both runs
# -----------------------------
plt.figure(figsize=(10, 6))
plt.plot(df["Time_min"], df["Run1_mol_l"], marker="o", linestyle="-", label="Run1 (low catalyst)")
plt.plot(df["Time_min"], df["Run2_mol_l"], marker="s", linestyle="-", label="Run2 (high catalyst)")

plt.title("Batch Reactor: Reactant Concentration vs Time")
plt.xlabel("Time (min)")
plt.ylabel("Concentration (mol/L)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

# -----------------------------
# d) Gas compression data
# -----------------------------
gas_data = {
    "Volume_L": [1, 2, 3, 4, 5],
    "Pressure_bar": [5.03, 2.52, 1.65, 1.26, 1.01]
}

df_gas = pd.DataFrame(gas_data)

# Calculate k = mean(P * V)
k = (df_gas["Pressure_bar"] * df_gas["Volume_L"]).mean()

print("\nGas constant k =", k)

# Scatter plot of measured data
plt.figure(figsize=(10, 6))
plt.scatter(df_gas["Volume_L"], df_gas["Pressure_bar"], label="Measured data", color="blue", s=60)

# Reference curve P = k/V
V_ref = np.linspace(1, 5, 200)
P_ref = k / V_ref
plt.plot(V_ref, P_ref, color="red", linewidth=2, label="Reference curve: P = k/V")

plt.title("Ethylene Gas Compression: Pressure vs Volume")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
