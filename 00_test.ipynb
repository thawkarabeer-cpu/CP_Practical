```python
# Student grade A - 95 marks, B - 85 marks, C - 75 marks, D - 65 marks, F - below 65 marks
# if elif else format
marks = int(input("Enter your marks: "))

if marks >= 95:
    grade = "A"
elif marks >= 85:
    grade = "B"
elif marks >= 75:
    grade = "C"
elif marks >= 65:
    grade = "D"
else:
    grade = "F"

print(f"Your grade is {grade}")
```
